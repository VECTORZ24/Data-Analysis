```python
import pandas as pd
pd.set_option('display.precision', 2)
```

```python
#Extract
raw_data=pd.read_csv("student_dropout_dataset.csv")
raw_data.head()
```

```python
#Transform
#Checking nan values
raw_data.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
fig1=sns.displot(raw_data["Family_Income"])
fig2=sns.displot(raw_data["Study_Hours_per_Day"])
fig3=sns.displot(raw_data["Stress_Index"])
plt.show()
```

```python
#The Family income distribution shows some outliers so we will fill nan values with the median value.
#For the stress_index and Study_Hours_Per_Day we will fill nan values with the mean value.
data=raw_data.copy()
data["Family_Income"]=data["Family_Income"].fillna(data["Family_Income"].median())
data["Stress_Index"]=data["Stress_Index"].fillna(data["Stress_Index"].mean())
data["Study_Hours_per_Day"]=data["Study_Hours_per_Day"].fillna(data["Study_Hours_per_Day"].mean())
```

```python
data.isnull().sum()
```

```python
#loading data (saving to csv files)
cleaned_data=data.copy()
cleaned_data.to_csv("Clean_Data.csv",index=False)
```
